```python
import pandas as pd
import os
```

```python
# ===== 1. 取值位置 =====
RAW_DIR = r"datas"
OUT     = r"out_data/allmonth.csv"
FILES  = ["2019-Oct.csv", "2019-Nov.csv", "2019-Dec.csv",
          "2020-Jan.csv", "2020-Feb.csv"]
first = True
total = 0

first = True
total = 0

for f in FILES:
    path = os.path.join(RAW_DIR, f)
    if not os.path.exists(path):
        print(f"跳过: {f}")
        continue

    for chunk in pd.read_csv(path, chunksize=1_000_000, low_memory=False):

        chunk["event_time"] = pd.to_datetime(chunk["event_time"], errors="coerce", utc=True).dt.tz_convert(None)
        chunk["date"]    = chunk["event_time"].dt.date
        chunk["hour"]    = chunk["event_time"].dt.hour
        chunk["weekday"] = chunk["event_time"].dt.weekday
        chunk["weekend"] = (chunk["weekday"] >= 5).astype("int8")

        chunk.to_csv(OUT, mode="w" if first else "a",
                     header=first, index=False, encoding="utf-8")
        first = False
        total += len(chunk)

    print(f"{f} 完成，累计 {total:,} 行")

print(f"总共 {total:,} 行 -> {OUT}")
```
